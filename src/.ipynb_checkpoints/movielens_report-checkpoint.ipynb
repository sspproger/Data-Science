## Класс Ratings
Класс `Ratings` отвечает за загрузку и первичную обработку данных из файла `ratings.csv`. При инициализации класса происходит чтение первых 1000 строк файла, после чего данные распределяются по соответствующим структурам для дальнейшего анализа. Класс также содержит статические методы для статистических вычислений.

**Атрибуты класса:**
* `userIds` — список идентификаторов пользователей
* `movieIds` — список идентификаторов фильмов
* `ratings_list` — список оценок (числовые значения)
* `timestamps` — список временных меток
* `years` — список годов, извлеченных из временных меток с использованием `datetime`
* `ratings_data` — список словарей с полной информацией о каждой оценке

**Статические методы:**
* `median(data)` — вычисляет медиану переданного списка чисел. Медиана — это значение, которое делит упорядоченный набор данных на две равные половины.

* `variance(data)` — вычисляет дисперсию (меру разброса) переданного списка чисел. Дисперсия показывает, насколько сильно значения отклоняются от среднего.

**Обработка ошибок:**
* `FileNotFoundError` — если файл не найден, выбрасывается исключение с соответствующим сообщением
* `Exception` — любые другие ошибки чтения файла также обрабатываются и пробрасываются дальше

Класс `Ratings` содержит два внутренних класса, каждый из которых отвечает за свой аспект анализа данных.

```python
%%timeit -r1 -n1
import movielens_analysis
ratings = movielens_analysis.Ratings('../datasets/ratings.csv')
all_ratings = ratings.ratings_list
median_all = movielens_analysis.Ratings.median(all_ratings)
print(f"Медиана всех оценок: {median_all}")
```

```python
%%timeit -r1 -n1
import movielens_analysis
ratings = movielens_analysis.Ratings('../datasets/ratings.csv')
all_ratings = ratings.ratings_list
variance_all = movielens_analysis.Ratings.variance(all_ratings)
print(f"Дисперсия всех оценок: {variance_all}")
```

3. Возможно, люди чаще оценивают фильмы, которые им понравились (чем те, что разочаровали)

**Дисперсия** же при этом составила примерно 1.38 что указывает на то что мнения пользователей умеренно расходятся

### Подкласс Movies
Подкласс `Movies` отвечает за анализ данных, связанных с фильмами. Для работы он использует данные оценок из класса `Ratings` и информацию о названиях фильмов из файла `movies.csv`. Все методы возвращают результаты в виде `OrderedDict`, что гарантирует сохранение порядка сортировки и удобочитаемость данных.

**Методы класса Movies:**

* `dist_by_year()` — подсчитывает количество оценок фильмов по годам. Результат сортируется по возрастанию годов. Позволяет увидеть, как менялась активность пользователей с течением времени и какие годы были наиболее "кинематографически активными". Использует предварительно извлеченное поле `year` из данных оценок.

* `dist_by_rating()` — подсчитывает, сколько раз встречалась каждая оценка (от 0.5 до 5.0) в датасете. Результат сортируется по возрастанию количества оценок. Помогает понять распределение пользовательских предпочтений — склонны ли пользователи ставить высокие или низкие оценки.

* `top_by_num_of_ratings(n)` — находит топ-N фильмов, получивших наибольшее количество оценок. Метод показывает, какие фильмы были самыми популярными с точки зрения активности пользователей, используя данные о названиях из `movies.csv`. Названия неизвестных фильмов помечаются как "Unknown {movieId}".

* `top_by_ratings(n, metric='average')` — находит топ-N фильмов с самыми высокими оценками. В качестве метрики можно использовать среднее арифметическое (`average`) или медиану (`median`). Позволяет увидеть фильмы с наилучшими пользовательскими рейтингами, исключая влияние выбросов при использовании медианы. Значения округляются до двух знаков после запятой.

* `top_controversial(n)` — находит самые противоречивые фильмы, то есть те, по поводу которых мнения зрителей сильно расходятся. В качестве меры разброса используется дисперсия оценок (метод `variance` из родительского класса). Чем выше дисперсия, тем больше разброс во мнениях. Этот метод помогает найти фильмы, которые либо гениальны, либо ужасны — в зависимости от точки зрения зрителя.


```python
%%timeit -r1 -n1
import movielens_analysis

ratings = movielens_analysis.Ratings('../datasets/ratings.csv')

movies_data = movielens_analysis.Movies('../datasets/movies.csv')

# Для controversial: Первые тысячу фильмов не соответствуют первым тысячам оценок.
# Оставляем только те оценки, для которых есть названия в первых 1000 фильмах,
# чтобы избежать появления "Unknown" в названиях фильмов.
# Для by_num_of_ratings, by_ratings_average, by_ratings_median 
# оставляем все данные, так как в результах нет "Unknown".

available_ids = set(movies_data.movie_titles.keys())
filtered_ratings = [r for r in ratings.ratings_data if r['movieId'] in available_ids]

ratings_movies_controversial = ratings.Movies(filtered_ratings, movies_data.movie_titles)
# Если нужно использовать все оценки, тогда используем ratings_movies,
# убираем фильтрацию и получаем Unknown+movie_id фильма в controversial.

ratings_movies = ratings.Movies(ratings.ratings_data, movies_data.movie_titles)

by_year = ratings_movies.dist_by_year()
by_rating = ratings_movies.dist_by_rating()
by_num_of_ratings = ratings_movies.top_by_num_of_ratings(5)
by_ratings_average = ratings_movies.top_by_ratings(n=5, metric='average')
by_ratings_median = ratings_movies.top_by_ratings(n=5, metric='median')
controversial = ratings_movies_controversial.top_controversial(5)

print("Сортировка оценок по годам:")
for key, value in by_year.items():
    print(f"  {key}: {value}")

print("\nСортировка оценок по рейтингу:")
for key, value in by_rating.items():
    print(f"  {key}: {value}")

print("\nТоп 5 фильмов, получивших наибольшее количество оценок:")
for title, count in by_num_of_ratings.items():
    print(f"  {title}: {count} оценок")

print("\nТоп 5 фильмов с самыми высокими оценками (по среднему арифметическому):")
for title, rating in by_ratings_average.items():
    print(f"  {title}: {rating}")

print("\nТоп 5 фильмов с самыми высокими оценками (по медиане):")
for title, rating in by_ratings_median.items():
    print(f"  {title}: {rating}")

print("\nТоп 5 самых противоречивых фильмов:")
for title, variance in controversial.items():
    print(f"  {title}: дисперсия {variance}")
```

   * Во все остальные года люди скучали без кино (но скорее всего в первой 1000 фильмов из датасета преобладают фильмы 1996, 2000, 2005 годов)

### Подкласс Users
Подкласс `Users` отвечает за анализ данных, связанных с пользователями. Он использует данные оценок из класса `Ratings` для изучения пользовательского поведения и выявления интересных паттернов.

**Методы класса Users:**

* `dist_by_num_of_ratings()` — подсчитывает количество оценок, поставленных каждым пользователем. Результат сортируется по убыванию активности. Позволяет выявить самых активных пользователей, которые внесли наибольший вклад в датасет.

* `dist_by_ratings(metric='average')` — анализирует, какие оценки в среднем ставит каждый пользователь. Можно использовать среднее арифметическое (`average`) или медиану (`median`). Помогает понять предпочтения различных групп пользователей — есть ли пользователи, которые стабильно ставят высокие оценки, или наоборот, строгие критики. Результаты округляются до двух знаков.

* `top_controversial(n)` — находит топ-N самых непредсказуемых пользователей, которые ставят и очень высокие, и очень низкие оценки. В качестве метрики используется дисперсия оценок конкретного пользователя. Чем выше дисперсия, тем более противоречивы оценки пользователя. Такие пользователи могут быть интересны для изучения, так как их сложно удовлетворить — они одинаково легко ставят как 0.5, так и 5.0.


```python
%%timeit -r1 -n1
import movielens_analysis

ratings = movielens_analysis.Ratings('../datasets/ratings.csv')

ratings_users = ratings.Users(ratings.ratings_data)

by_num_of_ratings = ratings_users.dist_by_num_of_ratings()
by_ratings_average = ratings_users.dist_by_ratings(metric='average')
by_ratings_median = ratings_users.dist_by_ratings(metric='median')
controversial = ratings_users.top_controversial(5)

print("Подсчет количества оценок поставленных каждым пользователем:")
for key, value in by_num_of_ratings.items():
    print(f"  {key}: {value}")

print("\nАнализирует какие оценки в среднем ставит пользователь (по средне арифметическому):")
for key, value in by_ratings_average.items():
    print(f"  {key}: {value}")

print("\nАнализирует какие оценки в среднем ставит пользователь (по медиане):")
for title, count in by_ratings_median.items():
    print(f"  {title}: {count}")

print("\nТоп 5 самых непредсказуемых пользователей (дисперсия):")
for title, rating in controversial.items():
    print(f"  {title}: {rating}")
```

# Класс Tags
Класс `Tags` предназначен для анализа пользовательских тегов из файла `tags.csv` датасета MovieLens. Теги — это короткие метки, которые пользователи присваивают фильмам (например, "sci-fi", "based on a book", "atmospheric").

**Основные возможности:**
- Загрузка первых 1000 тегов из файла
- Поиск самых "многословных" тегов (по количеству слов)
- Поиск самых длинных тегов (по количеству символов)
- Поиск пересечения этих двух топов
- Определение самых популярных тегов
- Поиск тегов по ключевому слову

Файл `tags.csv` содержит следующие колонки:
- `userId` — идентификатор пользователя
- `movieId` — идентификатор фильма
- `tag` — текстовый тег
- `timestamp` — время добавления

**Методы класса**

* `__init__(path_to_the_file)` — Загружает первые 1000 тегов из файла, приводит их к нижнему регистру.

* `most_words(n)` — Возвращает теги с наибольшим количеством слов в формате `{тег: количество_слов}`.

* `longest(n)` — Возвращает список самых длинных тегов по количеству символов.

* `most_words_and_longest(n)` — Возвращает теги, которые одновременно входят в топ-n по словам и по символам.

* `most_popular(n)` — Возвращает самые часто используемые теги в формате `{тег: частота}`.

* `tags_with(word)` — Возвращает все уникальные теги, содержащие указанное слово, отсортированные по алфавиту.


```python
%%timeit -r1 -n1
import movielens_analysis

tags = movielens_analysis.Tags('../datasets/tags.csv')

most_words_n = tags.most_words(3)
longest_n = tags.longest(5)
most_words_and_longest_n = tags.most_words_and_longest(3)
most_popular_n = tags.most_popular(5)
tags_with_result = tags.tags_with('funny')

print("3 тега с наибольшим количеством слов:")
for key, value in most_words_n.items():
    print(f"  {key}: {value}")

print("\n5 самых длинных тегов по количеству символов:")
for tag in longest_n: 
    print(f"  {tag}")

print("\nТеги, которые одновременно входят в топ-5 по словам и по символам:")
# most_words_and_longest_n возвращает список
for tag in most_words_and_longest_n:
    print(f"  {tag}")

print("\n5 самых часто используемых тегов:")
for tag, count in most_popular_n.items():
    print(f"  {tag}: {count}")

print("\nВсе уникальные теги, содержащие слово 'funny', отсортированные по алфавиту:")
for tag in tags_with_result: 
    print(f"  {tag}")
```

---

#### Выводы по Tags:
1) Пользователи используют теги по-разному:
   * Короткие жанровые теги (funny, action)
   * Длинные описательные теги (мини-рецензии)
2) Популярность vs Длина:
   * Короткие теги чаще используются (funny - 15 раз)
   * Длинные теги уникальны (встречаются по 1 разу)
3) Семантические группы:
   * Жанры: sci-fi, action, dark comedy
   * Оценки: funny, twist ending
   * Технические: oscar, assassin-in-training

# Класс Movies
Класс `Movies` отвечает за анализ данных из файла `movies.csv`. При инициализации он загружает первые 1000 строк файла, парсит их с учетом возможных запятых внутри названий фильмов и создает удобные структуры данных для дальнейшего анализа: список словарей с полной информацией, а также отдельные словари для быстрого доступа к названиям и жанрам по ID фильма.

**Методы класса Movies:**

* `dist_by_release()` — подсчитывает количество фильмов, выпущенных в каждом году. Год извлекается из названия фильма с помощью регулярного выражения, ищущего четыре цифры в круглых скобках в конце строки. Фильмы, у которых не удалось определить год выпуска, помечаются как 'Unknown'. Результат сортируется по убыванию количества фильмов, что позволяет увидеть, какие годы были наиболее продуктивными с точки зрения кинопроизводства.

* `dist_by_genres()` — подсчитывает частоту встречаемости каждого жанра в датасете. Учитывается, что один фильм может относиться к нескольким жанрам (разделитель |). Жанр (no genres listed) обрабатывается как отсутствие жанров. Результат сортируется по убыванию популярности, демонстрируя, какие жанры доминируют в выбранной выборке фильмов.

* `most_genres(n)` — находит топ-N фильмов с наибольшим количеством жанров. Для каждого фильма подсчитывается количество приписанных ему жанров, после чего фильмы сортируются по убыванию этого числа. При одинаковом количестве жанров применяется дополнительная сортировка по названию фильма в алфавитном порядке. Метод возвращает словарь, где ключами являются названия фильмов, а значениями — количество жанров, что позволяет выявить самые "жанрово-разнообразные" фильмы в выборке.



```python
%%timeit -r1 -n1
import movielens_analysis

mov = movielens_analysis.Movies('../datasets/movies.csv')

dist_by_release_n = mov.dist_by_release()
dist_by_genres_n = mov.dist_by_genres()
most_genres_n = mov.most_genres(5)

print("Количество фильмов, выпущенных в каждом году")
for key, value in dist_by_release_n.items():
    print(f"  {key}: {value}")

print("\nЧастота встречаемости каждого жанра в датасете")
for tag, count in dist_by_genres_n.items():
    print(f"  {tag}: {count}")

print("\nТоп 5 фильмов с наибольшим количеством жанров")
for tag, count in most_genres_n.items():
    print(f"  {tag}: {count}")
```

# Класс Links
Класс `Links` отвечает за анализ данных из файла `links.csv` и обогащение их информацией, полученной путем парсинга страниц IMDb. При инициализации он загружает первые 1000 строк файла, содержащих соответствия между movieId, IMDB ID и TMDB ID. Класс использует библиотеки `requests` для HTTP-запросов и `BeautifulSoup` для парсинга HTML-страниц IMDb, извлекая ключевую информацию о фильмах.

файл `links.csv` содержит следующие колонки:
* `movieId` — уникальный идентификатор фильма в локальной базе данных (соответствует `movieId` в файлах `movies.csv` и `ratings.csv`)
* `imdbId` — идентификатор фильма на платформе IMDb (Internet Movie Database). Используется для построения ссылок вида `https://www.imdb.com/title/tt{imdbId}/`
* `tmdbId` — идентификатор фильма на платформе TMDb (The Movie Database). Может содержать пропуски (значения `NaN`)

**Методы класса Links:**

* `_get_movie_info(imdb_id)` — внутренний метод для парсинга страницы фильма на IMDb. Извлекает следующую информацию:
  * `director` — имя режиссера фильма
  * `budget` — бюджет производства в долларах (целое число)
  * `gross` — мировые кассовые сборы в долларах (целое число)
  * `runtime` — длительность фильма в минутах
  
  Метод использует регулярные выражения для поиска и очистки числовых значений, обрабатывает временные форматы (часы и минуты) и возвращает словарь с полученными данными или `None` в случае ошибки.

* `top_directors(n)` — находит топ-N самых плодовитых режиссеров в выборке. Для каждого фильма из `links.csv` выполняет запрос к IMDb, извлекает имя режиссера и подсчитывает количество фильмов, снятых каждым режиссером. Результат сортируется по убыванию количества фильмов и возвращается в виде `OrderedDict`, где ключи — имена режиссеров, значения — количество фильмов.

* `most_expensive(n)` — находит топ-N самых дорогих фильмов по бюджету производства. Для каждого фильма извлекает значение бюджета со страницы IMDb и сортирует фильмы по убыванию бюджета. Возвращает `OrderedDict`, где ключи — `movieId`, значения — бюджет в долларах.

* `most_profitable(n)` — находит топ-N самых прибыльных фильмов. Прибыль рассчитывается как разница между мировыми кассовыми сборами (`gross`) и бюджетом производства (`budget`). Фильмы сортируются по убыванию прибыли. Возвращает `OrderedDict` с `movieId` в качестве ключей и прибылью в долларах в качестве значений.

* `top_cost_per_minute(n)` — находит топ-N фильмов с наибольшей стоимостью минуты экранного времени. Показатель рассчитывается как бюджет, разделенный на хронометраж (в минутах), и округляется до двух знаков после запятой. Фильмы с нулевым хронометражем исключаются из расчета. Результат сортируется по убыванию стоимости минуты и возвращается в виде `OrderedDict` с `movieId` и значением стоимости минуты.

```python
%%timeit -r1 -n1
import movielens_analysis

lin = movielens_analysis.Links('../datasets/links.csv')
movies = movielens_analysis.Movies('../datasets/movies.csv')

lin.links = lin.links[:20]  # ← ТОЛЬКО ПЕРВЫЕ 20 ФИЛЬМОВ!
print(f"Анализируем {len(lin.links)} фильмов")

top_directors_n = lin.top_directors(10)
most_expensive_n = lin.most_expensive(10)
most_profitable_n = lin.most_profitable(10)
top_cost_per_minute_n = lin.top_cost_per_minute(10)

print("\nТоп режиссеров:")
for director, count in top_directors_n.items():
    print(f"{director}: {count} фильмов")

print("\nТоп самых дорогих фильмов:")
for movie_id, budget in most_expensive_n.items():
    title = movies.movie_titles.get(int(movie_id), f"Unknown {movie_id}")
    print(f"{title}: ${budget:,}")

print("\nТоп самых прибыльных фильмов:")
for movie_id, profit in most_profitable_n.items():
    title = movies.movie_titles.get(int(movie_id), f"Unknown {movie_id}")
    print(f"{title}: ${profit:,}")

print("\nТоп фильмов по стоимости минуты:")
for movie_id, cost in top_cost_per_minute_n.items():
    title = movies.movie_titles.get(int(movie_id), f"Unknown {movie_id}")
    print(f"{title}: ${cost:,.2f} за минуту")
    
```

---

#### В результате анализа 20 фильмов 1995 года можно сделать несколько интересных наблюдений:

1) Самым дорогим фильмом оказался **Cutthroat Island** с бюджетом 98 млн долларов, за ним следуют **Money Train** (68 млн) и **Jumanji** (65 млн). Однако высокий бюджет не гарантирует успех — **Cutthroat Island** не вошел в список самых прибыльных фильмов, в отличие от Jumanji, который окупился почти в три раза.

2) По прибыли с большим отрывом лидирует **Toy Story** — 371 млн. долларов при бюджете около 30 млн. Это лучший показатель среди всех фильмов выборки. Второе место у **GoldenEye** с 292 млн, третье у **Jumanji** с 197 млн.

3) Интересно посмотреть на стоимость минуты экранного времени. Самые дорогие минуты оказались у **Cutthroat Island** (790 тыс. долларов за минуту), **Jumanji** (625 тыс.) и **Money Train** (618 тыс.). При этом **Toy Story** с относительно скромными 370 тыс. за минуту стал самым прибыльным фильмом.

4) Таким образом, анализ показывает, что успех фильма не всегда связан с размером бюджета. Анимация и проверенные франшизы показали лучшие результаты, в то время как некоторые дорогие проекты не смогли окупиться.
